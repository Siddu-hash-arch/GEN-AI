```python
import chromadb

client = chromadb.Client()

print("ChromaDB client created successfully!")
```

```python
collection = client.create_collection(name="my_first_collection")

print("Collection created:", collection.name)
```

```python
collection.add(
    documents=[
        "Python is a popular programming language for AI.",
        "The weather today is sunny and warm.",
        "Machine learning models learn patterns from data.",
        "I enjoy playing football on weekends.",
        "Deep learning uses neural networks with many layers.",
        "Paris is the capital city of France."
    ],
    ids=["doc1", "doc2", "doc3", "doc4", "doc5", "doc6"]
)

print("Documents added! Total documents:", collection.count())
```

```python
results = collection.query(
    query_texts=["Tell me about AI and neural networks"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f}  |  {doc}")
```

```python
results = collection.query(
    query_texts=["What is the capital of France?"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f}  |  {doc}")
```

```python
collection.add(
    documents=[
        "The new smartphone has an amazing camera.",
        "Stock prices rose sharply after the earnings report."
    ],
    metadatas=[
        {"category": "technology"},
        {"category": "finance"}
    ],
    ids=["doc7", "doc8"]
)

# Query only within the "technology" category
results = collection.query(
    query_texts=["Tell me about new gadgets"],
    n_results=2,
    where={"category": "technology"}
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f}  |  {doc}")
```

```python
from chromadb.utils import embedding_functions

hf_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

custom_collection = client.create_collection(
    name="custom_embedding_collection",
    embedding_function=hf_embedder
)

custom_collection.add(
    documents=[
        "I love dogs",
        "I like puppies",
        "The stock market crashed today"
    ],
    ids=["a1", "a2", "a3"]
)

results = custom_collection.query(
    query_texts=["puppy lovers"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f}  |  {doc}")
```

2. Add a `"category"` metadata field to every document, then filter a query by two different categories.

```python
try:
    client.delete_collection(name="exercise_meta_collection")
except Exception:
    pass

meta_collection = client.create_collection(name="exercise_meta_collection")
meta_collection.add(
    documents=[
        "The James Webb Space Telescope captures stunning images of distant galaxies.",
        "Baking sourdough bread requires patience and a healthy active starter.",
        "Quantum computing relies on qubits to perform complex calculations.",
        "Mount Everest is the highest mountain above sea level in the world.",
        "Learning to play the piano improves hand-eye coordination."
    ],
    metadatas=[
        {"category": "space"},
        {"category": "cooking"},
        {"category": "physics"},
        {"category": "geography"},
        {"category": "music"}
    ],
    ids=["m1", "m2", "m3", "m4", "m5"]
)
query_results = meta_collection.query(
    query_texts=["scientific topics"],
    n_results=3,
    where={
        "$or": [
            {"category": "space"},
            {"category": "physics"}
        ]
    }
)
for doc, metadata, distance in zip(query_results["documents"][0], query_results["metadatas"][0], query_results["distances"][0]):
    print(f"Category: {metadata['category']} | Distance: {distance:.4f} | {doc}")
```

3. Swap the embedding function to `'all-mpnet-base-v2'` and compare the distance scores to `all-MiniLM-L6-v2`.

```python
from chromadb.utils import embedding_functions
minilm_fn = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)
mpnet_fn = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-mpnet-base-v2"
)
docs = [
    "The James Webb Space Telescope captures stunning images of distant galaxies.",
    "Baking sourdough bread requires patience and a healthy active starter.",
    "Quantum computing relies on qubits to perform complex calculations.",
    "Mount Everest is the highest mountain above sea level in the world.",
    "Learning to play the piano improves hand-eye coordination."
]
ids = ["s1", "s2", "s3", "s4", "s5"]
query = "How are pictures of space taken?"
try:
    client.delete_collection("compare_minilm")
    client.delete_collection("compare_mpnet")
except Exception:
    pass

coll_minilm = client.create_collection("compare_minilm", embedding_function=minilm_fn)
coll_mpnet = client.create_collection("compare_mpnet", embedding_function=mpnet_fn)
coll_minilm.add(documents=docs, ids=ids)
coll_mpnet.add(documents=docs, ids=ids)
res_minilm = coll_minilm.query(query_texts=[query], n_results=5)
res_mpnet = coll_mpnet.query(query_texts=[query], n_results=5)
print(f"Query: '{query}'\n")
print(f"{'Document':<78} | {'all-MiniLM-L6-v2':<18} | {'all-mpnet-base-v2':<18}")
print("-" * 122)

minilm_map = {doc: dist for doc, dist in zip(res_minilm["documents"][0], res_minilm["distances"][0])}
mpnet_map = {doc: dist for doc, dist in zip(res_mpnet["documents"][0], res_mpnet["distances"][0])}

for doc in docs:
    mini_dist = minilm_map.get(doc, float('nan'))
    mpnet_dist = mpnet_map.get(doc, float('nan'))
    print(f"{doc:<78} | {mini_dist:<18.4f} | {mpnet_dist:<18.4f}")
```
